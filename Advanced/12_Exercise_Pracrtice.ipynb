```python
import ast
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv(r"C:\Users\HP\Documents\Learning Data Analytics\Python\Luke's Tutorial\Python_Data_Analytics_Course-main\data_jobs.csv")
```

```python
df['job_posted_date'] = pd.to_datetime(df['job_posted_date'])
```

```python
def clean_list(skill_list):
    if pd.notna(skill_list):
       return ast.literal_eval(skill_list)

df['job_skills'] = df['job_skills'].apply(clean_list)
df['job_skills']
```

```python
df_da = df[df['job_title_short'] == 'Data Analyst'].copy()
df_da
```

```python
df_da_explode = df_da.explode('job_skills')
df_da_explode['job_posted_month'] = df_da_explode['job_posted_date'].dt.strftime('%b')
df_da_explode['job_posted_month_no'] = df_da_explode['job_posted_date'].dt.month
```

```python
df_da_explode
```

```python
df_da_pivot = df_da_explode.pivot_table(index = 'job_posted_month_no', columns = 'job_skills', aggfunc = 'size', fill_value = 0)
```

```python
df_da_pivot.loc['Total'] = df_da_pivot.sum()
df_da_pivot
```

```python
df_da_pivot = df_da_pivot[df_da_pivot.loc['Total'].sort_values(ascending = False).index]
```

```python
df_da_pivot = df_da_pivot.drop('Total')
```

```python
df_da_pivot.iloc[:, :5].plot(kind = 'line')
plt.title('Top 5 skills for Data Analysts per Month')
plt.ylabel('Count')
plt.xlabel('')
plt.legend().set_visible(True)
plt.show()
```
