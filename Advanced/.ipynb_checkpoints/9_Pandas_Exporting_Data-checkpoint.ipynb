```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv(r"C:\Users\HP\Documents\Learning Data Analytics\Python\Luke's Tutorial\Python_Data_Analytics_Course-main\data_jobs.csv")
```

```python
df['job_posted_date'] = pd.to_datetime(df['job_posted_date'])
```

```python
df['job_posted_month'] = df['job_posted_date'].dt.strftime('%b')
```

```python
months = df['job_posted_month'].unique().tolist()

dict_month = {month: df[df['job_posted_month'] == month] for month in months}
dict_month['Feb']
```

```python
df_q1 = pd.concat([dict_month['Jan'],dict_month['Feb'],dict_month['Mar']], ignore_index = True)
df_q1
```
