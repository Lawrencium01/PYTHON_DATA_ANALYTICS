```python
import ast
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv(r"C:\Users\HP\Documents\Learning Data Analytics\Python\Luke's Tutorial\Python_Data_Analytics_Course-main\data_jobs.csv")
```

```python
df['job_posted_date'] = pd.to_datetime(df['job_posted_date'])
```

```python
def clean_list(skill_list):
    if pd.notna(skill_list):
       return ast.literal_eval(skill_list)

df['job_skills'] = df['job_skills'].apply(clean_list)
df['job_skills']
```

```python
df.head()
```

```python
df['job_work_from_home'].value_counts().plot(kind ='pie', startangle = 90, autopct = '%1.1f%%')
plt.title('Work from home status')
plt.ylabel('')
plt.show()
```

```python
df['job_title_short'].value_counts().plot(kind ='pie' )
plt.title('Job titles in luke dataset')
plt.ylabel('')
plt.show()
```

```python
df['job_title_short'].value_counts().plot(kind ='barh' )
plt.title('Job titles in luke dataset')
plt.gca().invert_yaxis()
plt.ylabel('')
plt.show()
```

```python
fig, ax = plt.subplots(1,3)

column_titles = ['job_work_from_home','job_no_degree_mention','job_health_insurance']

for i, column_title in enumerate(column_titles):
    df[column_title].value_counts().plot(kind = 'pie', ax = ax[i], title = column_title, startangle = 90, autopct = '%1.1f%%')
    ax[i].set_ylabel('')

plt.tight_layout()
```

```python
fig, ax = plt.subplots(1,3)

column_titles = {'job_work_from_home':'Work from home',
                 'job_no_degree_mention':'Jon Degree Req.',
                 'job_health_insurance':'Health Insurance Offered'
                }

for i, (column,title) in enumerate(column_titles.items()):
    df[column].value_counts().plot(kind = 'pie', ax = ax[i], title = title, startangle = 90, autopct = '%1.1f%%')
    ax[i].set_ylabel('')

plt.tight_layout()
```
