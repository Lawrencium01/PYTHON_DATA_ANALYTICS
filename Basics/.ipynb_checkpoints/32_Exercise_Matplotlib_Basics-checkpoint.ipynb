## Median Salary vs Data Science Jobs

```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv(r"C:\Users\HP\Documents\Learning Data Analytics\Python\Luke's Tutorial\Python_Data_Analytics_Course-main\data_jobs.csv")
```

```python
df['job_posted_date'] = pd.to_datetime(df['job_posted_date'])
```

```python
job_salary = df.groupby(by = 'job_title_short')['salary_year_avg'].median().sort_values(ascending = True)
```

```python
df.groupby(by = 'job_title_short')['salary_year_avg'].mean().sort_values(ascending = True)
```

```python
type(job_salary)
```

```python
plt.plot(job_salary.index, job_salary)
plt.xticks(rotation = 45, ha = 'right')
plt.show()
```

```python
job_salary.plot(kind = 'barh')
plt.title('Median Salary by Job Title')
plt.xlabel('Salary ($USD)')
plt.ylabel("")
plt.show()
```
