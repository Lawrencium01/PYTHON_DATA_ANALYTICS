```python
import random

# Base pool of common data science roles
base_titles = [
    "Data Scientist", "Senior Data Scientist", "Junior Data Scientist",
    "Machine Learning Engineer", "AI Researcher", "Data Analyst", 
    "Data Engineer", "Deep Learning Engineer", "NLP Scientist", 
    "Data Architect", "Quantitative Analyst", "Statistician",
    "Principal Data Scientist", "Data Science Manager", "Decision Scientist"
]

# Generate 100 titles by sampling with replacement (allows repeats)
data_science_titles = [random.choice(base_titles) for _ in range(100)]

# Output the result
for i, title in enumerate(data_science_titles, 1):
    print(f"{i}. {title}")
```

```python
analytic_list = []

for job in data_science_titles:
    if 'Data Analyst' in job:
        analytic_list.append(job)
analytic_list
```

```python
analytic_list = [job for job in data_science_titles if 'Data Analyst' in job]
analytic_list
```
