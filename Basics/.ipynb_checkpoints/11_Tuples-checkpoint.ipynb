```python
lukes_skill = ('python', 'sql', 'statistics', 'tableau')
print(id(lukes_skill))

new_skills = ('excel', 'power bi')
```

```python
lukes_skill  += new_skills
```

```python
lukes_skill
id(lukes_skill)
```

```python
lukes_skill = ['python', 'sql', 'statistics', 'tableau']
print(id(lukes_skill))

new_skills = ('excel', 'power bi')

lukes_skill  += new_skills

print(lukes_skill)
id(lukes_skill)
```

```python
tuple1 = ('mango', 'orange')
tuple2 = ('banana','cherry')

print(id(tuple1))
```

```python
tuple.__add__(tuple1,tuple2)
```

```python
tuple1 += tuple2
```

```python
id(tuple1)
```
