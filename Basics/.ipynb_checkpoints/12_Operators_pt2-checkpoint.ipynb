## Identity Operators

```python
salary_kelly = 100_000
salary_luke = 100_000

salary_kelly is salary_luke
```

```python
id(salary_kelly) == id(salary_luke)
```

```python
print(id(salary_kelly))
print(id(salary_luke))
```

```python
core_skills = ['Python', 'r', 'SQL']
lukes_skills = core_skills
```

```python
lukes_skills is core_skills
```
