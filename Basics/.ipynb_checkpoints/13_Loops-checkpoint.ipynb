## Loops

```python
numbers = [0,1,2,3,4,5]

for number in numbers:
    print(number)
```

```python
characters = 'PYTHON'

for character in characters:
    print(character)
```

```python
years_experience = {
    'luke': 3,
    'kelly': 6,
    'ken': 4,
    'Alex': 3
}
```

```python
for person in years_experience:
    print(person)
```

```python
for value in years_experience.values():
    print(value)
```

```python
for key,value in years_experience.items():
    print(key,value)
```

```python
for key,value in years_experience.items():
    if value >= 5:
        print(key,'has',value,'years of experience')
```

# While Loop

```python
while False:
    print('Data Nerd')
```

```python
number = 0
while (number < 5):
    print(number)
    number += 1
```

```python
count = 1
while (count < 7):
    print(count)
    count = count + 1
```

```python
years_experience = {
    'luke': 3,
    'kelly': 6,
    'ken': 4,
    'Alex': 3,
    'Mary': 7
}
```

```python
years = 0
index = 0

years_list = list(years_experience.items())
while(years < 5):
    key,years = years_list[index]
    if years >= 5:
        print(key,'has',years,'years of experience')
    index +=1
```
