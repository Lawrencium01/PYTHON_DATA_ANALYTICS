## User-Defined Functions

```python
base_salary = 100000
bonus_rate = 0.1

total_salary = base_salary * (1 + bonus_rate)
total_salary
```

```python
def calculate_salary():
    base_salary = 100000
    bonus_rate = 0.1
    
    total_salary = base_salary * (1 + bonus_rate)
    return total_salary
```

```python
calculate_salary()
```

```python
def calculate_salary(base_salary, bonus_rate = 0.1):
    total_salary = base_salary * (1 + bonus_rate)
    return total_salary
```

```python
salary1 = 100000
rate_1 = 0.1

calculate_salary(salary1,rate_1)
```
