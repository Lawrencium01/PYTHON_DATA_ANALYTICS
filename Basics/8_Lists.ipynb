# Unpacking

```python
job_skills = ['python','excel','sql','looker']
```

```python
skill1, skill2, skill3 = job_skills
```

```python
print(skill1)
print(skill2)
print(skill3)
```

```python
skill_concerned, *skill_dont_care = job_skills
```

```python
print(skill_concerned)
print(skill_dont_care)
```
