```python
# Create a list of 1,000,000 salaries ranging from 50,000 to 150,000
import random

salary_list = [random.randint(50000,100000) for _ in range(10000000)]
```

```python
import statistics
```

```python
%%timeit

statistics.mean(salary_list)
```

```python
import numpy as np
```

```python
%%timeit

np.mean(salary_list)
```
