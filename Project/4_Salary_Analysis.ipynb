# How well do jobs and skills pay for Data Analysts?

```python
import ast
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv(r"C:\Users\HP\Documents\Learning Data Analytics\Python\Luke's Tutorial\Python_Data_Analytics_Course-main\data_jobs.csv")
```

```python
df['job_posted_date'] = pd.to_datetime(df['job_posted_date'])
```

```python
def clean_list(skill_list):
    if pd.notna(skill_list):
       return ast.literal_eval(skill_list)

df['job_skills'] = df['job_skills'].apply(clean_list)
```

```python
df_US = df[df['job_country'] == 'United States'].dropna(subset = ['salary_year_avg'])
df_US
```

```python
job_titles = df_US['job_title_short'].value_counts().head(6).index.to_list()
job_titles
```

```python
df_US_top6 = df_US[df_US['job_title_short'].isin(job_titles)]
df_US_top6
```

```python
job_order = df_US_top6.groupby('job_title_short')['salary_year_avg'].median().sort_values(ascending = False).index
job_order
```

```python
sns.boxplot(data = df_US_top6, x = 'salary_year_avg', y = 'job_title_short', order = job_order)

plt.title('Salary Distribution in the United States')
plt.xlabel('Yearly Salary ($USD)')
plt.ylabel('')
ax = plt.gca()
ax.xaxis.set_major_formatter(plt.FuncFormatter(lambda x, pos: f"${int(x/1000)}K"))
plt.xlim(0,600000)
plt.show()
```

# Investigate Median Salary vs Skill for Data Analyst

```python
df_DA_US = df[(df['job_title_short'] == 'Data Analyst') & (df['job_country'] == 'United States')].copy()
df_DA_US = df_DA_US.dropna(subset = ['salary_year_avg'])
```

```python
df_DA_US = df_DA_US.explode('job_skills')
```

```python
df_DA_US[['salary_year_avg','job_skills']].head()
```

```python
df_DA_top_pay = df_DA_US.groupby('job_skills')['salary_year_avg'].agg(['count','median']).sort_values('median', ascending = False)
# df_DA_US.groupby('job_skills').agg(
#     'skill_count' = ('job_skills','count'),
#     'median_salary' = ('salary_year_avg','median')
# )
df_DA_top_pay = df_DA_top_pay.head(10)
```

```python
df_DA_skills = df_DA_US.groupby('job_skills')['salary_year_avg'].agg(['count','median']).sort_values('count', ascending = False)
df_DA_skills = df_DA_skills.head(10).sort_values('median', ascending = False)
df_DA_skills
```

```python
fig, ax = plt.subplots(2,1)

sns.set_theme(style = 'ticks')

# df_DA_top_pay.plot(kind = 'barh', y = 'median_salary', ax = ax[0])

sns.barplot(data = df_DA_top_pay, x = 'median', y = 'job_skills', ax = ax[0], hue = 'median', palette = 'dark:b_r')
ax[0].set_title('Top 10 Highest Paid Skills for Data Analysts')
ax[0].set_ylabel('')
ax[0].set_xlabel('')
# ax[0].invert_yaxis()
ax[0].legend().set_visible(False)
ax[0].xaxis.set_major_formatter(plt.FuncFormatter(lambda x, pos: f"${int(x/1000)}K"))

# df_DA_skills.plot(kind = 'barh', y = 'skill_count', ax = ax[1])

sns.barplot(data = df_DA_skills, x = 'median', y = 'job_skills', ax = ax[1], hue = 'median', palette = 'light:b')
ax[1].set_title('Top 10 Most In-Demand Skills for Data Analysts')
ax[1].set_ylabel('')
ax[1].set_xlabel('Median Salary ($USD)')
# ax[1].invert_yaxis()
ax[1].legend().set_visible(False)
#ax[1].legend().remove()
ax[1].xaxis.set_major_formatter(plt.FuncFormatter(lambda x, pos: f"${int(x/1000)}K"))

ax[1].set_xlim(0, 200000)
#ax[1].set_xlim(ax[0].get_xlim())
plt.tight_layout()
#fig.tight_layout()
plt.show()
```
